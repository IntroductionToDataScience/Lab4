### Exhaust Gas Data Quality

The Volvo Penta environmental testing department has provided a CSV file containing exhaust gas measurements collected during marine engine tests.

Unfortunately, the dataset contains several data quality problems, including missing values, duplicates, incorrect data types, and suspicious sensor readings.

Your job is to prepare the dataset for further analysis.

Your tasks:

1. Load *volvo_penta_exhaust_co2.csv* into a DataFrame named measurements. Display the data and examine its structure using info().

2. Identify missing values in each column. Check whether some apparently missing or invalid information has been stored as text instead of NaN.

3. Identify and remove exact duplicate records.

4. Convert Date to a proper date type and CO2Pct and ExhaustTempC to numerical types. Handle values that cannot be converted.

5. For this exercise, assume that valid CO₂ concentrations must be between 0% and 20%, and exhaust temperatures between 150°C and 650°C. Identify measurements outside these ranges and mark them as missing.

6. Count the remaining missing values. Replace them with the median of their respective numerical columns.

7. Verify that the resulting DataFrame has consistent data types, no missing values, and no exact duplicate records.

8. Sort the cleaned dataset by CO2Pct in descending order, reset its index, and export it to cleaned_exhaust_measurements.csv.

Think about it: One of the CO₂ measurements is unusually high but still within the accepted range. Should we automatically remove it? Explain your reasoning.

###  Marine Engine Emissions Data Integration

The Volvo Penta engineering department has collected results from several marine engine emission tests.

The information is stored in a SQLite database named volvo_penta_emission_tests.db, containing three tables:

| Table | Description |
|---|---|
| Engines | Engine identifiers, models, rated power, and vessel names. |
| TestSessions | Test dates, engine loads, and engine speeds. |
| GasReadings | NOₓ concentrations and exhaust gas temperatures recorded during the tests. |

**Hint: you can install Visual Studio Code extension to browse the db file and inspect the data first**

1. Connect to the SQLite database using sqlite3.

2. Load all three tables into separate Pandas DataFrames using pd.read_sql_query(). Examine their structures and identify the relationships between them.

3. Use TestSessions as the central dataset. Combine it with Engines using EngineID, then with GasReadings using TestID.

4. Inspect the resulting DataFrame. Explain why there are more rows than test sessions.

5. Identify and remove duplicated measurements. Remember that different ReadingID values do not necessarily mean the underlying measurements are different. For this exercise, each test should have only one measurement.

6. Convert TestDate to a date type and NOx_ppm to a numerical type, handling invalid text values appropriately.

7. For this exercise, assume that acceptable NOₓ concentrations are between 0 and 1200 ppm, and exhaust temperatures between 150°C and 650°C. Identify values outside these ranges and mark them as missing.

8. Count missing values and replace missing numerical measurements with their respective medians.

9. Verify the cleaned dataset, sort it by NOx_ppm in descending order, and reset its index.

10. Save the result as a new table named CleanedEmissionTests in the same SQLite database using to_sql(). Read the table back and verify the saved data.

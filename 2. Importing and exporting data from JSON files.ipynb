### Marine Engine Performance Analysis

The Volvo Penta engineering department has provided a JSON file containing telemetry data collected from several marine engines.

Each engine has its own information, including its model and vessel name, along with a list of daily measurements.

Unfortunately, the data is stored in a hierarchical structure, which is not particularly convenient for analysis.

Your job is to prepare it for the Data Science team.

#### Your tasks:

1. Load volvo_penta_engine_telemetry.json using Python's json module. Examine its structure.

2. Use pd.json_normalize() to transform the data into a DataFrame where each row represents a single daily measurement.

3. Include the engine identifier, vessel name, and engine model in every row.

4. Create a new column named FuelPerHour, calculated as fuelLiters / hours.

5. Sort the records by FuelPerHour in descending order.

6. Export the resulting DataFrame to engine_telemetry.csv without its index.

Question: Why does the resulting DataFrame contain more rows than the original JSON has top-level objects?

### Investigating Marine Engine Emissions

The Volvo Penta engineering department wants to investigate whether engine operating conditions are related to exhaust emissions.

Unfortunately, the required information comes from three separate systems, each exporting its data to a different JSON file.

The files contain engine specifications, daily operating measurements, and exhaust emission measurements.

Your task is to combine these datasets into a single DataFrame suitable for further analysis.

#### Your tasks:

1. Load all three JSON files into separate Pandas DataFrames. Examine their structure and identify the columns that can be used to combine them.

2. Use the engine operations dataset as the central DataFrame and merge it with the engine specifications using EngineID.

3. Merge the result with the exhaust emissions dataset using both EngineID and Date.

4. Verify that the resulting DataFrame contains 16 records and includes all the relevant information.

5. Sort the records by NOx_ppm in descending order and reset the DataFrame index to reflect the new row order.

6. Export the sorted DataFrame to engine_emissions_analysis.csv, including the new index.

A little Data Science question:

Examine the EngineLoadPct and NOx_ppm columns.

Do higher engine loads appear to be associated with higher NOₓ concentrations? What evidence can you find in the data?

And more importantly, can we conclude that increasing engine load always causes higher NOₓ emissions? Why or why not?
